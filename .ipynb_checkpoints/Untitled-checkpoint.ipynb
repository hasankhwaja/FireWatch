```python
def nextPermutation(nums: list[int]) -> None: 

    def generatePermutations(res, nums, idx): 

        # if index is at the end of the array 
        if idx == len(nums) - 1: 
            res.append(nums[:]) 
            return 
    
        # generate all permutations 
        for i in range(idx, len(nums)): 
            nums[idx], nums[i] = nums[i], nums[idx] 

            # Recur for next index 
            generatePermutations(res, nums, idx + 1) 

            # backtrack to restore original array 
            nums[idx], nums[i] = nums[i], nums[idx] 

    res = [] 
    generatePermutations(res, nums, 0) 

    # sort all permutations 
    res.sort() 

    # find curr permutation index 
    for i in range(len(res)): 
        # if curr permutaion matches input 
        if res[i] == nums: 
            # if not the last permutation 
            if i < len(res) - 1: 
                nums[:] = res[i + 1] 
            # if its the last permutation 
            else: 
                nums[:] = res[0]
            break 
        
if __name__ == "__main__":

    arr = [2, 4, 1, 7, 5, 0]
    
    nextPermutation(arr)
    
    for x in arr:
        print(x, end=" ")  
```

```python
def nextPermutation(nums: list[int]) -> None: 
    n = len(nums) 

    # find first decreasing element from the right 
    i = n - 2 
    while i >= 0 and nums[i] >= nums[i + 1]: 
        i -= 1 

    # if such element exists, swap it with thenuxt larger element 
    if i >= 0: 
        j = n - 1 
        while nums[j] <= nums[i]: 
            j -= 1 

    nums[i], nums[j] = nums[j], nums[i]

    # reverse the suffix 
    left = i + 1 
    right = n - 1 

    while left < right: 
        nums[left], nums[right] = nums[right], nums[left] 
        left += 1 
        right -= 1 
```

```python
def timeOfBuffering(arrivalRate, packets): 
    buffer = set() 
    n = len(packets) 

    total_seconds = (n + arrivalRate - 1) // arrivalRate 

    for t in range(1, total_seconds + 1): 
        start = (t - 1) * arrivalRate 
        end = min(t * arrivalRate, n) 

        arrivals = packets[start : end] 

        # check to play packet t
        if t in buffer: 
            buffer.remove(t) 
        elif t in arrivals: 
            # play packet t from arrivals 
            pass 
        else: 
            return t 

        # store future usable packts
        for packet in arrivals: 
            if packet > t: 
                buffer.add(packet) 

    return -1 
```
